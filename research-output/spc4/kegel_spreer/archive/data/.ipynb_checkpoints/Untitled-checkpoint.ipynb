```python
import csv
```

```python
friends=[]
with open('census_complete_list_verified_friends.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        friends.append(row)
```

```python
len(friends)
```

```python
friends[:10]
```

```python
import re

def extract_number(name):
    match = re.search(r'_(\d+)$', name)
    return int(match.group(1)) if match else float('inf')

# Natürliche Sortierung: zuerst nach K5/K6/K7, dann nach Zahl
friends = sorted(friends, key=lambda x: (x[0][:2], extract_number(x[0])))
```

```python
[x[0] for x in friends]
```

```python
def has_duplicates(lst):
    return len(lst) != len(set(lst))
```

```python
has_duplicates([x[0] for x in friends])
```

```python
seen = set()
unique_knots = []

for knot in friends:
    name = knot[0]
    if name not in seen:
        seen.add(name)
        unique_knots.append(knot)
```

```python
len(unique_knots)
```

```python
len(friends)
```

```python
friends=unique_knots
```

```python
[x[0] for x in friends]
```

```python
friends=[[x[0],eval(x[1])] for x in friends]
```

```python
friends[:4]
```

```python
len(friends)
```

```python
for x in friends:
    with open("census_friends.csv", "a") as output: #Opens output file # Use the a parameter to add a row
        csvwriter = csv.writer(output, delimiter = ",")
        csvwriter.writerow((x[0],x[1]))
```
